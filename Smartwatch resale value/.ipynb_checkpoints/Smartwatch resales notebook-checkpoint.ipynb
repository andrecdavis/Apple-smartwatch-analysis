```python
import pandas as pd

from sklearn.preprocessing import StandardScaler, MinMaxScaler, FunctionTransformer, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.model_selection import train_test_split, KFold, cross_val_score
from sklearn.feature_selection import SelectKBest

import numpy as np

from sklearn.linear_model import LinearRegression, Ridge

from sklearn.model_selection import GridSearchCV

from sklearn.metrics import confusion_matrix, accuracy_score, classification_report, mean_squared_error, r2_score
from sklearn.model_selection import cross_val_score

from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.svm import SVR
```

```python
def load_data():
    return pd.read_csv('Global_Smartwatch_Marketplace.csv')
```

```python
# We will first predict price based on everything except title

df = load_data()

df.columns = (
    df.columns
    .str.title()
    .str.replace('_Mm', '_mm', regex=False)
    .str.replace('_Id', '_ID')
)

# Replace NaNs in Case Size column with median

df['Case_Size_mm'] = df['Case_Size_mm'].fillna(df['Case_Size_mm'].median())


# There are price outliers, we will remove them here
Q1 = df['Price'].quantile(0.25)
Q3 = df['Price'].quantile(0.75)
IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df = df[(df['Price'] >= lower) & (df['Price'] <= upper)]
```

```python
# We'll treat the various values of condition, trying to rank them in a logical way

condition_0 = ['For parts or not working']
condition_1 = ['Good - Refurbished']
condition_2 = ['Used', 'Usado', '--']
condition_3 = ['Very Good - Refurbished']
condition_4 = ['Excellent - Refurbished', 'Open box',  'Certified - Refurbished', 'Caja abierta']
condition_5 = ['New', 'Nuevo', 'Novo', 'New other']

condition_list = [condition_0, condition_1, condition_2, condition_3, condition_4, condition_5]


replacement_dictionary = {x:i for i,lst in enumerate(condition_list) for x in lst}

df['Condition'].replace(replacement_dictionary, inplace=True)
```

```python
# Instead of one-hot, we will make new Brand_Freq, and Country_Freq columns

df['Brand'] = df['Brand'].map(df['Brand'].value_counts())
```

```python
df['Title'] = df['Title'].str.lower()
df['Title'] = df['Title'].str.replace(r'[^a-z0-9 ]', '', regex=True)

tfidf = TfidfVectorizer(
    max_features=100,     # limit size (important)
    ngram_range=(1,1),    # single words + pairs
    stop_words='english'
)



X = df.drop(columns = [ 
    'Price'
]
           )

y = df['Price']



X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=41)



X_train_tfidf = tfidf.fit_transform(X_train['Title'])
X_test_tfidf = tfidf.transform(X_test['Title'])

tfidf_columns = tfidf.get_feature_names_out()

X_train_tfidf_df = pd.DataFrame(
    X_train_tfidf.toarray(),
    columns=tfidf_columns,
    index=X_train.index
)

X_test_tfidf_df = pd.DataFrame(
    X_test_tfidf.toarray(),
    columns=tfidf_columns,
    index=X_test.index
)

X_train = X_train.drop(columns=['Title'])
X_test = X_test.drop(columns=['Title'])

X_train = X_train.join(X_train_tfidf_df)
X_test = X_test.join(X_test_tfidf_df)
```

```python
X_train.columns
```

```python
# Use one-hot to encode country


X_train[['Country']] = X_train[['Country']].fillna('Missing')


encoder = OneHotEncoder(handle_unknown='ignore')
encoded = encoder.fit_transform(X_train[['Country']].fillna('Missing'))

categorical_columns = [f'{col}_{cat}' for i, col in enumerate(X_train[['Country']].columns) 
                       for cat in encoder.categories_[i]]

one_hot_features = pd.DataFrame(encoded.toarray(), 
                                columns=categorical_columns,
                               index=X_train.index)

X_train = X_train.join(one_hot_features)

X_train = X_train.drop('Country', axis=1)


# Now encode the test data
encoded_test = encoder.transform(X_test[['Country']].fillna('Missing'))
categorical_columns_test = [f'{col}_{cat}' for i, col in enumerate(X_test[['Country']].columns) 
                       for cat in encoder.categories_[i]]

one_hot_features_test = pd.DataFrame(encoded_test.toarray(), 
                                columns=categorical_columns,
                               index=X_test.index)

X_test = X_test.join(one_hot_features_test)

X_test = X_test.drop('Country', axis=1)

```

```python
drop_columns = ['Brand', 'Condition', 'Case_Size_mm',]

X_train = X_train.drop(columns = drop_columns)
X_test = X_test.drop(columns = drop_columns)


# keep_columns = ['Brand', 'Condition', 'Case_Size_mm',
#        'Is_Worldwide_Shipping', 'aluminum', 'bluetooth',
#        'cellular', 'galaxy',  'garmin', 'gps',
#          'lte',
#         'series 10', 
#         'steel', 'titanium', 'ultra']

# X_train = X_train[keep_columns]
# X_test = X_test[keep_columns]
```

```python
train_data = X_train.copy()
train_data['Price'] = y_train


seller_avg = train_data.groupby('Seller_ID')['Price'].mean()

X_train['Seller_Avg_Price'] = X_train['Seller_ID'].map(seller_avg)
X_test['Seller_Avg_Price'] = X_test['Seller_ID'].map(seller_avg)

global_mean = y_train.mean()

X_test['Seller_Avg_Price'] = X_test['Seller_Avg_Price'].fillna(global_mean)

X_train = X_train.drop(columns=['Seller_ID'])
X_test = X_test.drop(columns=['Seller_ID'])
```

```python
lin = LinearRegression()
lin.fit(X_train, y_train)

y_pred_lin = lin.predict(X_test)


linreg_rmse = np.sqrt(mean_squared_error(y_test, y_pred_lin))
```

```python
linreg_rmse
```

```python
r2_score(y_test, y_pred_lin)
```

```python
rig = Ridge(alpha=0.7)
rig.fit(X_train, y_train)

y_pred_rig = rig.predict(X_test)
y_pred_rig = rig.predict(X_test)
```

```python
rigreg_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rig))
rigreg_rmse
```

```python
r2_score(y_test, y_pred_rig)
```

```python
svr = SVR(kernel='rbf', C=1.0, epsilon=0.1)
svr.fit(X_train, y_train)

# 4. Predict
y_pred_svr = svr.predict(X_test)
```

```python
svr_rmse = np.sqrt(mean_squared_error(y_test, y_pred_svr))
svr_rmse
```

```python
r2_score(y_test, y_pred_svr)
```
